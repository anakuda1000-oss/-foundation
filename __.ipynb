```python
import re
from collections import Counter

def analyze_text_structure(text):
    print("="*20)
    print("       文章構造の解析レポート")
    print("="*20)
    print()

    # 1. 全体量の把握
    total_chars = len(text)
    # 改行で分割して段落数をカウント
    paragraphs = [p for p in text.split('\n') if p.strip()]
    total_paragraphs = len(paragraphs)

    # 「。」や「！」などで文を分割して文の数をカウント
    sentences = [s for s in re.split(r'[。！？\n]', text) if s.strip()]
    total_sentences = len(sentences)

    # 平均の文の長さを計算
    avg_sentence_len = total_chars / total_sentences if total_sentences > 0 else 0

    print("【1. 文章の全体ボリューム】")
    print(f"- 総文字数    : {total_chars} 文字")
    print(f"- 段落数      : {total_paragraphs} 個")
    print(f"- 文の数      : {total_sentences} 文")
    print(f"- 1文あたりの平均長 : {avg_sentence_len:.1f} 文字")

    # 読みやすさの判定
    if avg_sentence_len > 60:
        print("  ※ 1文が長めです。箇条書きや「。」で区切るとさらに読みやすくなります。")
    else:
        print("  ※ 適切な文の長さで読みやすい構成です。")
    print("\n" + "-"*40 + "\n")

    # 2. 頻出キーワードの抽出
    # 2文字以上の日本語単語っぽいものを抽出する正規表現
    words = re.findall(r'[一-龥ぁ-んァ-ヶー]{2,}', text)

    # 一般的な助詞や接続詞などの除外リスト（ストップワード）
    stopwords = {'です', 'ます', 'こと', 'ため', 'よう', 'これ', 'それ', 'あれ', 'から', 'ので', 'また', 'しかし'}
    filtered_words = [w for w in words if w not in stopwords]

    # 単語の出現回数をカウント
    word_counts = Counter(filtered_words)
    top_words = word_counts.most_common(5)

    print("【2. 重要キーワード Top 5】")
    if top_words:
        for rank, (word, count) in enumerate(top_words, start=1):
            print(f" {rank}位: {word} ({count}回出現)")
    else:
        print(" 十分な長さのキーワードが見つかりませんでした。")

    print("\n" + "-"*40 + "\n")

    print("\n========================================")

# --- 試しに解析するサンプル文章 ---
sample_text = """
私は大学で英語を専攻しています。
そして大学に入ってから理系科目を学びなおしています。
具体的には大学教養の数学・物理、そしてアルゴリズムを勉強しています。
そしてpythonを使って基礎的なプログラムを動かしています。
"""

# プログラムを実行
analyze_text_structure(sample_text)
```
